```python
## Working with csv

import csv
with open('example.csv','w',newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name','city'])
    writer.writerow(['Inder','Sohna'])
with open('example.csv','r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
## Date and Time Operations

from datetime import datetime , timedelta
now = datetime.now()
print(now)
yesterday = now -timedelta(days=1)
print(yesterday)
```

That large number you're seeing is a timestamp.

Specifically, the value from time.time() represents the total number of seconds that have passed since a fixed point in time known as the Unix Epoch.

What is the Unix Epoch?
The Unix Epoch is January 1st, 1970, at 00:00:00 UTC (Coordinated Universal Time).

So, the number 1761132390.9365766 means that over 1.76 billion seconds have elapsed between January 1st, 1970, and the moment that line of code was executed.
