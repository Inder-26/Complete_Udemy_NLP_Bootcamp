## Decison Tree Classifier Implementation 

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
```

```python
iris
```

```python
print(iris['DESCR'])
```

```python
print(iris['target'])
```

```python
## Independent Features
X=pd.DataFrame(iris['data'],columns=['sepal length in cm','sepal width in cm',
                                   'petal length in cm','petal width in cm'])
```

```python
## Dependent Features
y=iris['target']
```

```python
## Train Test Split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X, y, test_size=0.2,
                                                random_state=10)
```

```python
## Applying Decison Tree Classifier
from sklearn.tree import DecisionTreeClassifier
tree_classifier=DecisionTreeClassifier()
```

```python
tree_classifier.fit(X_train,y_train)
```

```python
## Visualize Decison Tree
from sklearn import tree
plt.figure(figsize=(18,18))
tree.plot_tree(tree_classifier,filled=True)
```

```python
post_prune_tree_classifier=DecisionTreeClassifier(max_depth=2)
```

```python
post_prune_tree_classifier.fit(X_train,y_train)
```

```python
## Visualize Decison Tree After Post Pruning
from sklearn import tree
plt.figure(figsize=(10,10))
tree.plot_tree(post_prune_tree_classifier,filled=True)
```

```python
y_pred_tree=tree_classifier.predict(X_test)
y_pred_post_prune=post_prune_tree_classifier.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix,classification_report
```

```python
cm1=confusion_matrix(y_test,y_pred_tree)
cm2=confusion_matrix(y_test,y_pred_post_prune)
class_report1=classification_report(y_test,y_pred_tree)
class_report2=classification_report(y_test,y_pred_post_prune)
print("-"*56)
print("Confusion Matrix without post pruning :\n",cm1)
print("-"*56)
print("Confusion Matrix with post pruning with max_depth =2:\n",cm2)
print("-"*56)
print("Classification Report without post pruning :\n",class_report1)
print("-"*56)
print("Classification Report with post pruning with max_depth:\n",class_report2)
print("-"*56)
```

## Decision Tree Classifier Preprunning and Hyperparameter Tuning

```python
params={
    'criterion':['gini','entropy','log_loss'],
    'splitter':['best','random'],
    'max_depth':[1,2,3,4,5],
    'max_features':['sqrt','log2']
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
tree_classifier
grid=GridSearchCV(tree_classifier,param_grid=params,cv=5,scoring='accuracy')
```

```python
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred_grid=grid.predict(X_test)
```

```python
cm3=confusion_matrix(y_test,y_pred_grid)
class_report3=classification_report(y_test,y_pred_grid)
print("-"*56)
print("Confusion Matrix with pre pruning :\n",cm3)
print("-"*56)
print("Classification Report with pre pruning :\n",class_report3)
print("-"*56)
```

```python
from sklearn.metrics import accuracy_score
score=accuracy_score(y_test,y_pred_grid)
print('Accuracy Score :',score)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Note: In a true Jupyter/Colab environment, you would import 'display' 
# from IPython.display, but since this environment is restricted, 
# we'll use print() for the table and just rely on plt.show() for the plot.

# Data (Only key comparison metrics for plotting)
data = {
    'Metric': ['Accuracy', 'Macro Avg F1-Score', 'Weighted Avg F1-Score'],
    'No Pruning (Original Tree)': [0.97, 0.96, 0.97],
    'Post-pruning (max_depth=2)': [0.93, 0.93, 0.93],
    'Pre-pruning (Grid Search)': [1.00, 1.00, 1.00]
}

df = pd.DataFrame(data).set_index('Metric').T
models = df.index.tolist()

fig, ax = plt.subplots(figsize=(10, 6))

x = np.arange(len(models)) 
width = 0.25 

# Plot the three metrics for each model group
rects1 = ax.bar(x - width, df['Accuracy'], width, label='Accuracy', color='#1f77b4')
rects2 = ax.bar(x, df['Macro Avg F1-Score'], width, label='Macro Avg F1-Score', color='#ff7f0e')
rects3 = ax.bar(x + width, df['Weighted Avg F1-Score'], width, label='Weighted Avg F1-Score', color='#2ca02c')
rects = [rects1, rects2, rects3]

# Function to add text labels on the bars
def autolabel(rects):
    for rect in rects:
        height = rect.get_height()
        ax.annotate(f'{height:.2f}', 
                    xy=(rect.get_x() + rect.get_width() / 2, height),
                    xytext=(0, 3), 
                    textcoords="offset points",
                    ha='center', va='bottom',
                    fontsize=9)

# Apply labels to all bar sets
for rect_set in rects:
    autolabel(rect_set)

# Set X-axis labels to model names
ax.set_xticks(x)
ax.set_xticklabels(models, rotation=0)

# Set Y-axis and title
ax.set_ylabel('Score')
ax.set_title('Model Performance Comparison')
ax.set_ylim(0.9, 1.01)


ax.legend(loc='upper left', bbox_to_anchor=(1.01, 1.0), fontsize=10) 

# Adjust the plot area (left, bottom, width, height) to ensure the legend fits
plt.subplots_adjust(right=0.85) # This shifts the right edge of the plot to the left
plt.show()


# --- Styled Table (Plain print version for reference) ---
table_data = {
    'Metric': ['Accuracy', 'Macro Avg Precision', 'Macro Avg Recall', 'Macro Avg F1-Score', 'Weighted Avg Precision', 'Weighted Avg Recall', 'Weighted Avg F1-Score'],
    'No Pruning (Original Tree)': [0.97, 0.96, 0.97, 0.96, 0.97, 0.97, 0.97],
    'Post-pruning (max_depth=2)': [0.93, 0.93, 0.95, 0.93, 0.95, 0.93, 0.93],
    'Pre-pruning (Grid Search)': [1.00, 1.00, 1.00, 1.00, 1.00, 1.00, 1.00]
}
table_df = pd.DataFrame(table_data).set_index('Metric')
print("\n---------------------------------- Model Performance Comparison Table -----------------------------------")
print(table_df.to_string(float_format="{:.2f}".format))
```
