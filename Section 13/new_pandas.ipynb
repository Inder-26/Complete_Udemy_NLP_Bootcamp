```python
import pandas as pd
sales_data_df = pd.read_csv('sales_data.csv')
sales_data_df.head()
```

```python
sales_data_df.info()
```

```python
total_sales_by_product = sales_data_df.groupby('Product Category')['Total Revenue'].sum()
total_sales_by_product
```

```python
total_sales_by_product.plot(xlabel='Product Category',ylabel='Total Revenue',kind='bar', color='teal')
```

```python
## Plot sales trend over time
sales_trend=sales_data_df.groupby('Date')['Total Revenue'].sum().reset_index()
sales_trend
```

```python
plt.plot(sales_trend['Date'],sales_trend['Total Revenue'])
```

```python
# Convert to datetime
sales_data_df['Date'] = pd.to_datetime(sales_data_df['Date'])

# Group by week and sum revenue
weekly_sales = sales_data_df.resample('W', on='Date')['Total Revenue'].sum().reset_index()

# See weekly data
weekly_sales.head()

# Plot weekly revenue trend
plt.figure(figsize=(12,5))
plt.plot(weekly_sales['Date'], weekly_sales['Total Revenue'])
plt.title("Weekly Revenue Trend")
plt.xlabel("Week")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
# Convert Date to datetime if not already
sales_data_df['Date'] = pd.to_datetime(sales_data_df['Date'])

# Group by month
monthly_sales = sales_data_df.resample('M', on='Date')['Total Revenue'].sum().reset_index()

# Plot monthly trend
plt.figure(figsize=(12,5))
plt.plot(monthly_sales['Date'], monthly_sales['Total Revenue'])
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Total Revenue")
plt.xticks(rotation=45)
plt.show()
```

```python
weekly_sales['Moving_Avg'] = weekly_sales['Total Revenue'].rolling(window=4).mean()

plt.figure(figsize=(12,5))
plt.plot(weekly_sales['Date'], weekly_sales['Total Revenue'], label="Weekly Revenue")
plt.plot(weekly_sales['Date'], weekly_sales['Moving_Avg'], label="4-Week Moving Avg")
plt.title("Weekly Revenue with Smooth Trend")
plt.xlabel("Week")
plt.ylabel("Revenue")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
sales_data_df['Date'] = pd.to_datetime(sales_data_df['Date'])

# Weekly
weekly = sales_data_df.resample('W', on='Date')['Total Revenue'].sum().reset_index()

# Monthly
monthly = sales_data_df.resample('M', on='Date')['Total Revenue'].sum().reset_index()

plt.figure(figsize=(12,5))
plt.plot(weekly['Date'], weekly['Total Revenue'], label="Weekly Revenue")
plt.plot(monthly['Date'], monthly['Total Revenue'], label="Monthly Revenue", linewidth=3)
plt.title("Weekly vs Monthly Revenue Trend")
plt.xlabel("Time")
plt.ylabel("Revenue")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
weekly['Forecast'] = weekly['Total Revenue'].rolling(window=4).mean()

plt.figure(figsize=(12,5))
plt.plot(weekly['Date'], weekly['Total Revenue'], label="Actual")
plt.plot(weekly['Date'], weekly['Forecast'], label="4 Week Forecast", linestyle='--')
plt.title("Weekly Revenue Forecast")
plt.xlabel("Week")
plt.ylabel("Revenue")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
fig, axs = plt.subplots(2, 2, figsize=(15,10))

# Line weekly
axs[0,0].plot(weekly['Date'], weekly['Total Revenue'])
axs[0,0].set_title("Weekly Revenue")

# Line monthly
axs[0,1].plot(monthly['Date'], monthly['Total Revenue'])
axs[0,1].set_title("Monthly Revenue")

# Bar weekly
axs[1,0].bar(weekly['Date'], weekly['Total Revenue'])
axs[1,0].set_title("Weekly Revenue Bar Chart")

# Moving Average
weekly['MA'] = weekly['Total Revenue'].rolling(4).mean()
axs[1,1].plot(weekly['Date'], weekly['Total Revenue'])
axs[1,1].plot(weekly['Date'], weekly['MA'])
axs[1,1].set_title("Weekly Revenue + Moving Average")

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
print(tips)
```

```python
import matplotlib.pyplot as plt
sns.scatterplot(x='total_bill', y='tip', data=tips)
plt.title('Scatter Plot of Total Bill vs Tip')
plt.show()
```

```python
sns.lineplot(x='size', y='total_bill', data=tips)
plt.title('Line Plot of Total Bill by Size')
plt.show()
```

Categorical Plots

```python
sns.barplot(x='day', y='total_bill', data=tips)
plt.title('Line Plot of Total Bill by Day')
plt.show()
```

```python
## Box plot

sns.boxplot(x='day', y='total_bill', data=tips)
plt.title('Box Plot of Total Bill by Day')
plt.show()
```

```python
## Violin plot
sns.violinplot(x='day', y='total_bill', data=tips)
plt.title('Violin Plot of Total Bill by Day')
plt.show()
```

```python
## Histogram

sns.histplot(tips['total_bill'],bins=10,kde=False)
```

```python
sns.histplot(tips['total_bill'],bins=10,kde=True)
```

```python
## KDE Plot
sns.kdeplot(tips['total_bill'],fill=True)
```

```python
## Pair plot

sns.pairplot(tips)
```

```python
## Heatmap
corr=tips[['total_bill','tip','size']].corr()
corr
```

```python
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
