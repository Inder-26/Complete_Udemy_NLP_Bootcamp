## Handling Imbalanced Dataset

### Upsampling: Increasing the number of data points in the minority class.
### Downsampling: Reducing the number of data points in the majority class.

```python
import numpy as np
import pandas as pd

np.random.seed(0)

n_samples = 1000
class_zero_ratio = 0.9

n_class_0 = int(n_samples * class_zero_ratio)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0,n_class_1
```

```python
## Create DATAFRAME with imbalanced dataset
class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc=2, scale=1, size=n_class_1),
    'feature_2': np.random.normal(loc=2, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})
```

```python
df = pd.concat([class_0,class_1]).reset_index(drop=True)
```

```python
df.tail()
```

```python
df['target'].value_counts()
```

## Upsampling

```python
df_minortity=df[df['target']==1]
df_majority=df[df['target']==0]
```

```python
from sklearn.utils import resample

df_minortity_upsampled = resample(df_minortity,replace=True,
         n_samples=len(df_majority),
         random_state=42)
```

```python
df_minortity_upsampled.shape
```

```python
df_minortity_upsampled.head()
```

```python
df_upsampled = pd.concat([df_majority,df_minortity_upsampled])
```

```python
df_upsampled['target'].value_counts()
```
