```python
## Writing multiple lines to a file
lines = ['This is my first line.\n','I think I am writing this in a list.\n','It will use a loop and a writelines function to write all entries at once.\n']
with open('example.txt', 'a') as file:
    file.writelines(lines)
with open('example.txt','r') as file:
    for line in file:
        print(line.strip())
```

```python
## Counting Lines,Words, and Characters in a file

def count_file_stats(filename):
    with open(filename,'r') as file:
        lines = file.readlines()
        num_lines = len(lines)
        num_words = sum(len(line.split()) for line in lines)
        num_chars = sum(len(line) for line in lines)
    print(f"The no of lines in the file are :{num_lines} \n")
    print(f"The no of words in the file are :{num_words} \n")
    print(f"The no of chars in the file are :{num_chars} \n") 
count_file_stats('example.txt')
```

Working With File Paths

```python
## Creating a new directory

import os
new_directory = 'new'
os.mkdir(new_directory)
print(f"Directory '{new_directory}' is created.")
```

```python
## Listing files and directories

items = os.listdir(".")
print(items)
```

```python
## Joining Paths

directory_name ='folder'
file_name ='file.txt'
full_path =os.path.join(directory_name, file_name)
print(full_path)
```

```python
## Full Path

import os
full_path = os.path.join(os.getcwd())
print(full_path)

full_path = os.path.join(os.getcwd(),directory_name,file_name)
print(full_path)
```

```python
## Checking if a File Exists

example_file = 'example2.txt'
if os.path.exists(example_file):
    print(f"The path {example_file} exists.")
else:
    print(f"The path {example_file} does not exist.")
```

```python
## Checking If a Path Is a File or Directory

path = 'new'
if os.path.isfile(path):
    print(f"The path '{path}' is a file.")
elif os.path.isdir(path):
    print(f"The path '{path}' is a dirctory.")
else:
    print(f"The path {path} is neither a file nor a directory.")
```

```python
## Relative and Absolute path
relative_path = 'example.txt'
absolute_path = os.path.abspath(relative_path)
print(absolute_path)
```
