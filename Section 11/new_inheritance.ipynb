Inheritance is a fundamental concept in object-oriented programming that allows a class to inherit attributes and methods from another class. In this article, we will discuss how to implement inheritance in Python, covering both single and multiple inheritance with examples.

Inheritance is derived from real-world scenarios. For example, you inherit property from your parents, such as a house or land, but the reverse is not possible. Similarly, in Python, inheritance works in the same way, where a child class inherits properties from a parent class.

Single Inheritance

```python
class Car:
    def __init__(self, windows, doors, enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype

    def drive(self):
        print(f"Person will drive the {self.enginetype} car")

car1 = Car(6,6,'Ev')
car1.drive()
```

```python
class Tata(Car):
    def __init__(self, windows, doors, enginetype, ADAS):
        super().__init__(windows, doors, enginetype)
        self.ADAS = ADAS

    def sefdrive(self):
        print(f"Tata supports ADAS :{self.ADAS}")
    
    def __str__(self):
        return f"Casr has {self.windows} windows , {self.doors} doors , engine type is {self.enginetype} and supports ADAS :{self.ADAS}"
```

```python
Tata_Harrier = Tata(4,4,'EV',True)
print(Tata_Harrier)
```

Multiple Inheritance

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print("Subclasses must implement this method.")
```

```python
class Pet:
    def __init__(self, owner):
        self.owner = owner
```

```python
class Dog(Animal, Pet):
    def __init__(self, name, owner):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)

    def speak(self):
        return f"{self.name} says woof"
```

```python
dog = Dog("Buddy", "Inder")
print(dog.speak())
print(f"Owner: {dog.owner}")
```

Encapsulation

```python
## Access modifiers

## Public

class Example:
    def __init__(self):
        self.data = "I'm public!"     # Public attribute

    def show(self):
        print(self.data)             # Public method

obj = Example()
print(obj.data)                      # Direct access (Allowed)
obj.show()                           # Direct access (Allowed)
```

```python
## Protected (._variable)

class Example:
    def __init__(self):
        self._data = "I'm protected!"   # Protected attribute

    def _show(self):
        print(self._data)               # Protected method

class Derived(Example):
    def reveal(self):
        self._show()

        print(self._data)               # Access in subclass (Allowed)

obj = Derived()
print('\n')
print(obj._data)
print('\n')                        # Direct access (Allowed but discouraged)
obj._show()
print('\n')                             # Direct access (Allowed but discouraged)
obj.reveal()
```

```python
## Private

class Example:
    def __init__(self):
        self.__data = "I'm private!"      # Private attribute

    def __show(self):
        print(self.__data)                # Private method

    def expose(self):
        self.__show()                     # Access private within class

obj = Example()
# print(obj.__data)                      # Error! Not accessible directly
obj.expose()                             # Allowed: method inside class accesses private

# Name Mangling (Not recommended, but possible)
print(obj._Example__data)                # Allowed: access with name mangling
```

```python
# Public Version
class DemoPublic:
    def __init__(self):
        self.data = "Public"

# Protected Version
class DemoProtected:
    def __init__(self):
        self._data = "Protected"

# Private Version
class DemoPrivate:
    def __init__(self):
        self.__data = "Private"
```

```python
class Person:
    def __init__(self, name, age, gender):
        self._name = name  # protected variable
        self._age = age    # protected variable
        self.gender = gender

class Employee(Person):
    def __init__(self, name, age, gender):
        super().__init__(name, age, gender)

employee = Employee("Chris", 34, "male")
print(employee._name)  # Output: Chris
```

```python
class Student:
    def __init__(self, name, student_id):
        self.name = name         # Public attribute
        self.__student_id = student_id  # Private attribute

    def get_student_id(self):
        # Accessing private attribute inside the class (allowed)
        return self.__student_id

# Create an object
s = Student("Inder", 118)

print(s.name)           # Allowed, public attribute

# print(s.__student_id) # Error! AttributeError: 'Student' object has no attribute '__student_id'

print(s.get_student_id()) # Allowed, using a method inside the class

# Name Mangling (not recommended)
print(s._Student__student_id) # Allowed, but breaks encapsulation
```
