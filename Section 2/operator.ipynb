```python
a=10
b=6

add_result = a + b
sub_result = a - b
mul_result = a * b
div_result = a / b
floor_div_result = a // b
modulo_result = a % b
exponetiation_result = a ** b

print("The sum of a and b is:", add_result)
print("The difference of a and b is:", sub_result)
print("The product of a and b is:", mul_result)
print("The division of a by b is:", div_result)
print("The floor division of a by b is:", floor_div_result)
print("The modulo of a by b is:", modulo_result)
print("The exponentiation of a to the power of b is:", exponetiation_result)
```

```python
#String operations
str1 = "Hello"
str2 = "World"
concat_result = str1 + " " + str2
print("The concatenated string is:", concat_result)

repeat_result = str1 * 3
print("The repeated string is:", repeat_result)

#Comparison operations
is_equal = (a == b)
is_not_equal = (a != b)
is_greater = (a > b)
is_less = (a < b)
is_greater_equal = (a >= b)
is_less_equal = (a <= b)
print("Is a equal to b?:", is_equal)
print("Is a not equal to b?:", is_not_equal)
print("Is a greater than b?:", is_greater)
print("Is a less than b?:", is_less)
print("Is a greater than or equal to b?:", is_greater_equal)
print("Is a less than or equal to b?:", is_less_equal)
print("\n\n")

#Logical operations
logical_and = (a > 5) and (b < 10)
logical_or = (a < 5) or (b < 10)
logical_not = not(a > b)
print("Logical AND result:", logical_and)
print("Logical OR result:", logical_or)
print("Logical NOT result:", logical_not)
```
