```python
## real use case
def read_large_file(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line
```

### Closure Function
#### A closure is a function defined inside another function, which can access variables from the enclosing function's scope. This is similar to concepts in other languages like C# and Java.

```python
def main_welcome():
    message = "Welcome"
    def sub_welcome():
        print("Welcome to the Advanced Python course")
        print("Please learn this concepts properly")
        print(message)
    return sub_welcome
```

```python
closure_func = main_welcome()
closure_func()
```

```python
def main_welcome(func, lst):
    def sub_welcome():
        print("Welcome to the Advanced Python course")
        print(func(lst))
    return sub_welcome
```

```python
closure_func = main_welcome(len, [1,2,3,4,5])
closure_func()
```

```python
def main_welcome(print_message, func, lst):
    def sub_welcome():
        print(print_message) # This now uses the built-in print()
        print(func(lst))
    return sub_welcome

#closure_func = main_welcome('Trying to execute the print()', len, 'list together \n',len, [1,2,3,4,5])
#closure_func()

# Pass the last 3 items as one tuple for the 'lst' argument
closure_func = main_welcome('Trying to execute the print()', len, ('list together \n', len, [1,2,3,4,5]))

closure_func()
```

```python
def main_welcome(print_message, func, lst):
    def sub_welcome():
        print(print_message)
        
        # Access the first item in the tuple
        print(lst[0]) 
        
        # You could still run the function on the whole list if you want
        print(f"The length of the tuple is: {func(lst)}") 
        
    return sub_welcome

closure_func = main_welcome('Trying to execute the print()', len, ('list together \n', len, [1,2,3,4,5]))
closure_func()
```

## Introduction to Decorators

### Decorators are a powerful and flexible feature in Python that allow you to modify the behavior of a function or class method. They are commonly used to add functionality to functions or methods without modifying the actual code.

```python
def main_welcome(func):
    def sub_welcome():
        print("Welcome to the Advanced Python course")
        func()
        print("Please learn this concepts properly")
    return sub_welcome
```

```python
@main_welcome
def course_introduction():
    print("This is an advanced Python course")
```

```python
course_introduction()
```

## Another Example: Decorator with Pre- and Post-Functionality

### Decorators can be used to execute code before and after the decorated function is called.

```python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello")
```

```python
say_hello()
```

## Decorators with Arguments

### Decorators can also accept arguments, allowing for more flexible and dynamic behavior.

```python
def repeat(n):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(n):
                func(*args, **kwargs)
        return wrapper
    return decorator
```

```python
@repeat(6)
def say_hello():
    print("using decorators with positional arguments and keyword arguments")
```

```python
say_hello()
```
