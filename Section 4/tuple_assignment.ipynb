
### Assignment 7: Unpacking Tuples

Create a tuple with 5 elements and unpack it into 5 variables. Print the variables.

```python
numbers = (1, 2, 3, 4, 5, 6)
a, b, c, d, e, f= numbers
print("Packed tuples :")
print(a)
print(b)
print(c)
print(d)
print(e)
print(f)

print("\nUnpacked tuples :")

a,b,c,d,e,f = numbers
print(a,b,c,d,e,f)
```
