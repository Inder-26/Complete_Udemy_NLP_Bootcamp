```python
student ={"name":"inder","age":21,"state":"haryana","grade":'A+'}
print(student)
```

```python
#Accessing dictionary
print(student['grade'])
print(student['name'])

#using get() method
print(student.get('state'))
```

```python
#modifying dictionaries
student['age']= 31

student['state']= 'Punjab'
print(student)
```

```python
# use to delete a key-value pair
del student['grade']
print(student)
```

```python
#Common Dictionary Methods

#    .keys() returns all keys.
#    .values() returns all values.
#    .items() returns all key-value pairs as tuples.
keys = student.keys()
print(keys)
values = student.values()
print(values)
items = student.items()
print(items)
```

```python
#Shallow Copy of Dictionaries
student_copy = student
print(student)
print(student_copy)
student['name'] = 'inder'
print(student)
print(student_copy)
```

```python
student_copy1 = student.copy()
print(student_copy1)
student['name'] = 'Inderjeet'
print(student)
print(student_copy1)
```

```python
## Iterating over dictionaries
for key in student.keys():
    print(key)

for value in student.values():
    print(value)

for key,value in student.items():
    print(f"{key}:{value}")
```

```python
## nested dictionaries
students = {
    'student1': {'name': 'Inderjeet', 'age': 21},
    'student2': {'name': 'Pital', 'age': 22}
}
print(students)

## Access in nested dictionary
print("\n")
print(students['student2']['name'])
print(students['student2']['age'])
```

```python
for student_id, student_info in students.items():
    print(f"{student_id}: {student_info}")
    for key, value in student_info.items():
        print(f"{key}: {value}")
```

```python
## Dictionary comprehension
evens={x: x**2 for x in range(20) if x%2==0}
print(evens)
```

```python
## Practical Example: Counting Frequency of Elements in a List

## You can use a dictionary to count the frequency of elements in a list.

numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
frequency = {}
for number in numbers:
    if number in frequency:
        frequency[number] += 1
    else:
        frequency[number] = 1
print(frequency)
```

```python
## Merging Two Dictionaries

## You can merge two dictionaries using the double asterisk ** operator.

dict1={'a':1,'b':2}
dict2={'c':3,'d':4}
merged_dict = {**dict1, **dict2}
print(merged_dict)
```
